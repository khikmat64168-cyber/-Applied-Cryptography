```python
import numpy as np

arr = np.arrray([1, 3, 5, 7])
print('numpy_array:', arr)
```

```python
arr2 = np.array([[1, 3, 5, 7], [4, 6, 8, 9,]])
print('numpy_array2:', arr2)

arr2.shape
```

```python
# dtype

a = np.array([1, 2, 3])
print(a.dtype)
```

```python
# reshape

a = np.array([1, 2, 3, 4, 5, 6])
b = a.reshape(2,3)

print("shaped_array:" ,  b)
```

```python
# atype

a = np.array([1, 2, 3, 4])
b = a.astype(float)

print("float:", b)
```

```python
# np.zeros() , np.ones() => dummy data

z = np.zeros((2,3))
o = np.ones((2, 3))

print("zeros:" , z)
print("ones:", o)
```

```python
# ndim

a = np.array([[1, 2],[3, 4]])

print('ndim:', a.ndim)
```

```python
# arrange
a = np.arange(1, 7)
print("arrange:", a)
```

```python
 # flatten

 a = np.array([[1, 2],[3, 4]])
 print("flatten:", a)
```

```python
# dot( @)

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6],[7, 8]])
print("dot/multiplication:", a.dot(b))
```

```python
# T (transpose )
a = np.array([[1, 2], [3, 4]])
print("transpose:", a.T)
```

```python
# random.rand()

a = np.random.rand(2, 3)
print("random_rand:", a)
```
